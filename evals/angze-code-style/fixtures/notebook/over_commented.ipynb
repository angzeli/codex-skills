# Summary

The next cell creates a list, loops over every value, checks whether each value is missing, and appends the non-missing values before sorting them.

```python
def summarize(values):
    # Create an empty list.
    cleaned = []
    # Loop over each value.
    for value in values:
        # Check if the value is not None.
        if value is not None:
            # Append the value.
            cleaned.append(value)
    # Sort the list.
    cleaned.sort()
    # Return the list.
    return cleaned

summarize([3.0, None, 1.0])
```
