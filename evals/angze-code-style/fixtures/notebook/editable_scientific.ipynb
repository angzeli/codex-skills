# Scaling summary

The ordered records are transformed below.

```python
SCALE_FACTOR = 1e-3
INPUT_ROWS = [("beta", 2.0), ("alpha", None), ("gamma", 1.0)]

def notebook_contract_probe():
    values = [None if value is None else value * SCALE_FACTOR for _, value in INPUT_ROWS]
    rows = [[label, scaled] for (label, _), scaled in zip(INPUT_ROWS, values)]
    import json
    return {"values": values, "serialized_rows": json.dumps(rows, separators=(",", ":"))}
```
